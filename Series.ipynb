### **Pandas Series** ---> A **Pandas Series** is a **one-dimensional labeled array** that stores data in a **single column**, similar to a column in **Excel** or **Google Sheets**, where each value has an associated **index (label)**.

### Pandas Series is a column of data in Table like in Excel or Google Sheets with labels called as indexed attached to each value.

```python
pd.Series([40,50,60],index=["Bread","Milk","Butter"],name="Price of Grocery Items")
```

### **Series** ---> A **Pandas Series** is a **one-dimensional (1-D) labeled array** that can store **any type of data**, such as **integers, floating-point numbers, strings (objects), booleans, or mixed data types**.

#### **Features of a Series**
- **1-D Array** ---> Stores data in a single column.
- **Indexed** ---> Each value has an associated index (label).
- **Supports Multiple Data Types** ---> Can store integers, floats, strings, booleans, or mixed data.

#### Data Type ---> Integers -----> Data in a Series(Column) ---> Quantity of items in the stock
#### Data Type ---> Float -----> Data in a Series(Column) ---> Price or Product Ratings
#### Data Type ---> Strings -----> Column ---> Product Name
#### Data Type ---> Boolean -----> Column ---> Is item in Stock ?
#### Data Type ---> Dates/ times -----> Column---> Expiry Dates or Sale Dates
#### Mixed Types ---> Milk,40,True

### **Common Data Types in a Pandas Series**

#### **Integer (`int`)** ---> Used to store **whole numbers**, such as the **quantity of items in stock**.

#### **Float (`float`)** ---> Used to store **decimal numbers**, such as **product prices** or **ratings**.

#### **String (`str` or `object`)** ---> Used to store **text data**, such as **product names** or **customer names**.

#### **Boolean (`bool`)** ---> Used to store **True/False** values, such as **whether an item is in stock**.

#### **Date/Time (`datetime`)** ---> Used to store **dates and times**, such as **expiry dates**, **sale dates**, or **order dates**.

#### **Mixed Data Types** ---> A Series can also store **different types of data together**, for example: `"Milk"`, `40`, `True`.

## **Mathematical Functions in Pandas**

#### **1. `sum()`** ---> Used to calculate the **sum of all values** present in a Series.

#### **Syntax** ---> `series.sum()`

---

#### **2. `product()`** ---> Used to calculate the **product (multiplication)** of all values present in a Series.

#### **Syntax** ---> `series.product()`

---

#### **3. `mean()`** ---> Used to calculate the **average (mean)** of all values present in a Series.

#### **Syntax** ---> `series.mean()`

---

#### **4. `median()`** ---> Used to calculate the **median (middle value)** of all values present in a Series.
##### Note :---> Median also provides the idea about the average of the data. For coming up with median, we need to SORT the value in Ascending manner and then the middle value after sorting the data 

#### **Syntax** ---> `series.median()`

---

#### **5. `mode()`** ---> Used to calculate the **most frequently occurring value(s)** in a Series.

#### **Syntax** ---> `series.mode()`

---

#### **6. `std()`** ---> Used to calculate the **standard deviation** of the values present in a Series.
##### Note :---> If i need to get the idea that how the values inside the series are distributed from the average or mean value then we can come up with the standard deviation.

#### **Syntax** ---> `series.std()`

---

#### **7. `Variance [var()]`** ---> Used to calculate the **variance** of the values present in a Series.
##### Note :---> Variance also rpvides the insight about the distribution of the data from the mean value and variance is the square of the standard deviation
##### varaince = std ** 2

#### **Syntax** ---> `series.var()`

---

#### **8. `min()`** ---> Used to find the **minimum value** present in a Series.

#### **Syntax** ---> `series.min()`

---

#### **9. `max()`** ---> Used to find the **maximum value** present in a Series.

#### **Syntax** ---> `series.max()`

---

#### **10. `describe()`** ---> Used to display a **statistical summary** of the numerical values in a Series.
##### Note :---> describe fn provides the idea about the summary of the important mathematical calculations for the data

#### **Syntax** ---> `series.describe()`

##### **`describe()` Includes:**
- **count** ---> Total number of non-null values.
- **mean** ---> Average of all values.
- **std** ---> Standard deviation of the values.

- **75%** ---> Third Quartile (75th Percentile).
- **max** ---> Largest value in the Series.

#### **11. `cumsum()`** ---> Used to calculate the **cumulative sum** of the values in a Series.

#### **Syntax** ---> `series.cumsum()`

#### **12. `cumprod()`** ---> Used to calculate the **cumulative product** of the values in a Series.

#### **Syntax** ---> `series.cumprod()`

#### **13. `abs()`** ---> Used to return the **absolute (positive)** value of each element in a Series.

#### **Syntax** ---> `series.abs()`

---

#### **14. `round()`** ---> Used to round the values in a Series to the specified number of decimal places.

#### **Syntax** ---> `series.round(decimals)`

### **Indexing in a Pandas Series** *(To access one or more values from a Series)*

#### **1. Positive Indexing** ---> Works just like Python lists. Indexing starts from **0** and moves from **left to right**.

##### Note --> NO Negative Indexing

#### **Syntax** ---> `series[index]`

---

#### **2. Negative Indexing** ---> Indexing starts from **-1** and moves from **right to left**.

#### **Syntax** ---> `series[-index]`

---

#### **3. Label-Based Indexing** ---> Access values using their **index labels** instead of position.

#### **Syntax** ---> `series[label]`

---

#### **4. Multiple Value Indexing** ---> Access multiple values by passing a **list of indices or labels**.

#### **Syntax** ---> `series[[index1, index2]]` or `series[[label1, label2]]`

---

#### **5. Slicing** ---> Used to access a **range of values** from a Series using the slicing operator (`:`).

#### **Syntax** ---> `series[start:stop:step]`

#### **Note** ---> The **start index is included**, while the **stop index is excluded**.

---

#### **6. Label-Based Slicing** ---> Used to access a **range of values using index labels**.

#### **Syntax** ---> `series["start_label":"end_label"]`

#### **Note** ---> In **label-based slicing**, **both the start and end labels are included**.

---

#### **7. Fancy Indexing** ---> Used to access **multiple non-consecutive values** from a Pandas Series by passing a **list of index positions or labels**.

#### **Syntax (Position-Based)** ---> `series[[index1, index2, index3]]`

#### **Syntax (Label-Based)** ---> `series[[label1, label2, label3]]`
