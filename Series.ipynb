```python
import numpy as np
import pandas as pd
```

### **Pandas Series** ---> A **Pandas Series** is a **one-dimensional labeled array** that stores data in a **single column**, similar to a column in **Excel** or **Google Sheets**, where each value has an associated **index (label)**.

### Pandas Series is a column of data in Table like in Excel or Google Sheets with labels called as indexed attached to each value.

```python
pd.Series([40,50,60],index=["Bread","Milk","Butter"],name="Price of Grocery Items")
```

### **Series** ---> A **Pandas Series** is a **one-dimensional (1-D) labeled array** that can store **any type of data**, such as **integers, floating-point numbers, strings (objects), booleans, or mixed data types**.

#### **Features of a Series**
- **1-D Array** ---> Stores data in a single column.
- **Indexed** ---> Each value has an associated index (label).
- **Supports Multiple Data Types** ---> Can store integers, floats, strings, booleans, or mixed data.

#### Data Type ---> Integers -----> Data in a Series(Column) ---> Quantity of items in the stock
#### Data Type ---> Float -----> Data in a Series(Column) ---> Price or Product Ratings
#### Data Type ---> Strings -----> Column ---> Product Name
#### Data Type ---> Boolean -----> Column ---> Is item in Stock ?
#### Data Type ---> Dates/ times -----> Column---> Expiry Dates or Sale Dates
#### Mixed Types ---> Milk,40,True

### **Common Data Types in a Pandas Series**

#### **Integer (`int`)** ---> Used to store **whole numbers**, such as the **quantity of items in stock**.

#### **Float (`float`)** ---> Used to store **decimal numbers**, such as **product prices** or **ratings**.

#### **String (`str` or `object`)** ---> Used to store **text data**, such as **product names** or **customer names**.

#### **Boolean (`bool`)** ---> Used to store **True/False** values, such as **whether an item is in stock**.

#### **Date/Time (`datetime`)** ---> Used to store **dates and times**, such as **expiry dates**, **sale dates**, or **order dates**.

#### **Mixed Data Types** ---> A Series can also store **different types of data together**, for example: `"Milk"`, `40`, `True`.

### Creating a Series in Multiple Ways
#### 1. Lists
#### 2. Numpy Array
#### 3. Dictionary
#### 4. Custom Index
#### 5. Scalar Value

```python
## Series From List

country = ['India', 'China', 'USA', 'Nepal']

pd.Series(country)
```

```python
batsman_runs=[48,64,72,98,114,183]

pd.Series(batsman_runs,dtype='int32')
```

```python
subject_name=['Maths','Physics','Chemisry','Biology','English']
subject_marks=[90,93,64,90,98]

pd.Series(data=subject_marks,index=subject_name,name='Subject Marks data',dtype='int32')
```

```python
### creating an series using a Numpy Array
values=np.arange(10,21)
pd.Series(values)
```

### **Dictionary to Series**

#### A **Dictionary** is a Python data structure that stores data in **key-value pairs**.

#### When a **dictionary is converted into a Pandas Series**, the **keys become the index (labels)** and the **values become the data**.

#### Since the dictionary already contains **keys**, we **do not need to explicitly specify the index**.

#### **Syntax** ---> `pd.Series(dictionary)`
## ------------------------ Notes ------------------------
#### Distionary to Series

##### Dictionary is another Data Structure that stores data in a key value pair and just beacuse it stores the data in a key value pair so we don't have to explicitly mention the index

```python
data_items={'Maths':78,'English':92,'Chemistry':64}
pd.Series(data_items)
```

```python
## custom index to Series
values=[1,2,3,4,5]
pd.Series(data=values,index=['a1','a2','a3','a4','a5'])
```

```python
## Scalar Value inside an Series

pd.Series(4,index=['a','b','c','d','e'])
```
