## Data Exploration ---> Attributes & Functions for Dataframe

### **`Attributes`**

#### **1. shape** ---> It provides information about the **number of rows and columns** present in the DataFrame.
#### **Syntax** ---> `df.shape`

#### **2. dtypes** ---> It provides the **data type of each column** in the DataFrame.
#### **Syntax** ---> `df.dtypes`

#### **3. index** ---> It provides the **index labels** of the DataFrame.
#### **Syntax** ---> `df.index`

#### **4. columns** ---> It provides the **names of all the columns** present in the DataFrame.
#### **Syntax** ---> `df.columns`

#### **5. values** ---> It provides the **values stored inside the DataFrame** as an array. (it always returns a 2D numpy array representing the value of column row by row)
#### **Syntax** ---> `df.values`

#### **6. ndim** ---> It provides the **number of dimensions** of the DataFrame.
#### **Syntax** ---> `df.ndim`

#### **7. size** ---> It provides the **total number of elements** present in the DataFrame.
#### **Syntax** ---> `df.size`

#### **8. T** ---> It provides the **transpose** of the DataFrame, where rows become columns and columns become rows.
#### **Syntax** ---> `df.T`

### **Functions**

#### **1. head()** ---> It provides the **first 5 rows** of the DataFrame by default. (if i want that instead of top 5 rows i want to see fixed number of rowws from the rows, that number i can pass inside the head function parameter)
#### **Syntax** ---> `df.head()` and `df.nead(Number)`

#### **2. tail()** ---> It provides the **last 5 rows** of the DataFrame by default.
#### **Syntax** ---> `df.tail()` and `df.tail(Number)`

#### **3. sample()** ---> It is used to **randomly select rows** from the DataFrame.
#### **Syntax** ---> `df.sample()` and `df.tail(sample)`

#### **4. info()** ---> It provides information about the **DataFrame, columns, data types, and memory usage**.
#### **Syntax** ---> `df.info()`

#### **5. describe()** ---> It provides a **statistical summary** of the numerical columns.
#### **Syntax** ---> `df.describe()`

#### **6. sum()** ---> It calculates the **sum of values** in the DataFrame.
#### **Syntax** ---> `df.sum()`

#### **7. mean()** ---> It calculates the **mean (average)** of the values.
#### **Syntax** ---> `df.mean()`

#### **8. max()** ---> It provides the **maximum value** from the DataFrame.
#### **Syntax** ---> `df.max()`

#### **9. min()** ---> It provides the **minimum value** from the DataFrame.
#### **Syntax** ---> `df.min()`

#### **10. count()** ---> It counts the **non-empty values** in each column.
#### **Syntax** ---> `df.count()`

#### **11. sort_values()** ---> It is used to **sort the DataFrame based on a particular column**.
#### **Syntax** ---> `df.sort_values("column_name")`

#### **12. drop()** ---> It is used to **remove rows or columns** from the DataFrame.
#### **Syntax** ---> `df.drop("column_name", axis=1)`

#### **13. value_counts()** ---> It counts the **number of occurrences of each unique value** in a column.
#### **Syntax** ---> `df["column_name"].value_counts()`

#### **14. isnull()** ---> It checks for **missing values** and returns `True` or `False`.
#### **Syntax** ---> `df.isnull()`

#### **15. dropna()** ---> It is used to **remove rows or columns containing missing values**.
#### **Syntax** ---> `df.dropna()`

#### **16. fillna()** ---> It is used to **fill missing values** with a specified value.
#### **Syntax** ---> `df.fillna(value)`

### **Mathematical Functions in DataFrame**

#### **1. `sum()`** ---> It calculates the **sum of all values** present in a column or in multiple columns.
#### **Syntax** ---> `df.sum()`

#### **2. `mean()`** ---> It calculates the **average (mean)** of the values present in a column or multiple columns.
#### **Syntax** ---> `df.mean()`

#### **3. `min()`** ---> It provides the **minimum value** present in a column or multiple columns.
#### **Syntax** ---> `df.min()`

#### **4. `max()`** ---> It provides the **maximum value** present in a column or multiple columns.
#### **Syntax** ---> `df.max()`

#### **5. `median()`** ---> It calculates the **middle value (median)** of the values present in a column or multiple columns.
#### **Syntax** ---> `df.median()`

#### **6. `std()`** ---> It calculates the **standard deviation** of the values.
#### **Syntax** ---> `df.std()`

#### **7. `var()`** ---> It calculates the **variance** of the values (of the columns).
#### **Syntax** ---> `df.var()`

#### **8. `count()`** ---> It counts the **number of non-missing values** present in a column or multiple columns.
#### **Syntax** ---> `df.count()`

#### **9. `prod()`** ---> It calculates the **product of all values** present in a column or multiple columns.
#### **Syntax** ---> `df.prod()`

### **Selecting Specific Rows from a DataFrame**

#### **`loc[]`** ---> It is used to select rows or columns using their **labels (names)**. (it fetches the record as per the labelled location of the value)

#### **Syntax** ---> `df.loc[row_label]`

#### **`iloc[]`** ---> It is used to select rows or columns using their **integer positions (index numbers)**. (it fetches the record as per the integer index location of the value)

#### **Syntax** ---> `df.iloc[row_index]`

#### **Fancy Indexing**

#### **Fancy Indexing** ---> It is used to **select multiple specific rows or columns at the same time** by providing a list of index labels or positions.

#### **Using `loc[]`** ---> Select multiple rows using their **labels**.

#### **Syntax** ---> `df.loc[[row_1, row_2, row_3]]`

#### **Using `iloc[]`** ---> Select multiple rows using their **integer positions**.

#### **Syntax** ---> `df.iloc[[0, 2, 4]]`

```python
df.loc[[1,4,7,8,13]]
```

```python
data.loc[['Abhi','Sumit']]
```
