### **Attributes in a Pandas Series**

#### **1. `.values`** ---> Returns only the values of the Series as a NumPy array.

#### **Syntax** ---> `series.values`

---

#### **2. `.index`** ---> Returns the index (labels) of the Series. ( `returns only the index values from the given column` ).

#### **Syntax** ---> `series.index`

---

#### **3. `.dtype`** ---> Returns the data type of the values stored in the Series. (`Dtype provides the idea about the data type of the values that you are storing`)

#### **Syntax** ---> `series.dtype`

---

#### **4. `.shape`** ---> Returns the shape (number of rows) of the Series.(`This attribute provides the idea about the row, column information`)

#### **Syntax** ---> `series.shape`

---

#### **5. `.size`** ---> Returns the total number of elements in the Series. (`This attributes provides the idea about the total number of rows inside the dataset`)

#### **Syntax** ---> `series.size`

---

#### **6. `.name`** ---> Returns or sets the name of the Series. (`This attribute provides the idea about name of the series`)

#### **Syntax** ---> `series.name`

---

#### **7. `.index.name`** ---> Returns or sets the name of the index.

#### **Syntax** ---> `series.index.name`

---

#### **8. `.ndim`** ---> Returns the number of dimensions of the Series. (`This attribute provides the idea about the dimension of the given dataset`)

#### **Syntax** ---> `series.ndim`

---

#### **9. `.empty`** ---> Returns `True` if the Series is empty; otherwise returns `False`.

#### **Syntax** ---> `series.empty`

---

#### **10. `.hasnans`** ---> Returns `True` if the Series contains missing (`NaN`) values; otherwise returns `False`.

#### **Syntax** ---> `series.hasnans`

---

#### **11. `.is_unique`** ---> Returns `True` if all values in the Series are unique; otherwise returns `False`. (`This provides us with the idea about wwhether if unique values are their or not inside the column`)

#### **Syntax** ---> `series.is_unique`

##### **Note** ---> It is used to check whether the Series contains any duplicate values.

### **Series Methods**

#### **1. `head()`** ---> Returns the **top 5 rows** of a Series or DataFrame by default. It is useful for getting a quick preview of the dataset. If you want to display a specific number of rows, you can pass that number as an argument.

#### **Syntax** ---> `series.head()` or `series.head(n)`

##### **Example** ---> `series.head(10)` displays the first 10 rows, and `series.head(7)` displays the first 7 rows.

----

#### **2. `tail()`** ---> Returns the last **5 rows** of a Series or DataFrame by default. It is useful for previewing the end of the dataset.

#### **Syntax** ---> `series.tail()` or `series.tail(n)`

##### **Example** ---> `series.tail(10)` displays the bottom 10 rows, and `series.tail(7)` displays the bottom 7 rows.

----

#### **3. `sample()`** ---> Returns one or more random rows from a Series. It is useful for randomly inspecting the data. (`sample() provides any custom row from the dataset for us to preview it`)

#### **Syntax** ---> `series.sample()` or `series.sample(n)`

----

#### **4. `value_counts()`** ---> Counts the frequency of each unique value present in the Series. (`its provides the idea about the count of the values (categorical values), if i am having a column where the categorical value count i want to calculate then i can go with value_counts()`)

#### **Syntax** ---> `series.value_counts()`

----

#### **5. `sort_values()`** ---> Sorts the values of the Series in ascending order by default (`this method sort the value in an ascending manner. It can be useful for us for the numerical columns and also can even sort the value in descending manner as well).

#### **Syntax** ---> `series.sort_values()`

##### **Note** ---> Use `ascending=False` to sort in descending order.

----

#### **6. `sort_index()`** ---> Sorts the Series based on its index labels. (`this method sorts the index of the dataset`)

#### **Syntax** ---> `series.sort_index()`

----

#### **7. `unique()`** ---> Returns all the unique values present in the Series.

#### **Syntax** ---> `series.unique()`

----

#### **8. `nunique()`** ---> Returns the total number of unique values in the Series.

#### **Syntax** ---> `series.nunique()`

----

#### **9. `count()`** ---> Returns the total number of non-missing (non-NaN) values in the Series.

#### **Syntax** ---> `series.count()`

----

#### **10. `copy()`** ---> Creates a copy of the Series so that changes made to the new Series do not affect the original Series.

#### **Syntax** ---> `series.copy()`

### **Method Chaining** ---> Method chaining is the process of using multiple methods together in a single statement. The output of the first method becomes the input for the next method, and the output of the second method becomes the input for the third method, and so on, until the required result is obtained.

#### **Syntax** ---> `object.method1().method2().method3()`

##### **`Note: -`** We will use multiple methods together to get the reuired value in way the first method's output will be second method input then 2nd method output will be 3rd method input

```python
avg.sort_values(ascending=False).head(1)
```

```python
avg.sort_values(ascending=False).head(1).values[0]
```
